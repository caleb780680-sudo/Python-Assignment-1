```python
# Task 1: String Concatenation

# Part A: Concatenating String 1 ("Hello ") and String 2 (user input name)
string1 = "Hello "
name = input("Enter your Name: ")
concatenated_string = string1 + name
print(concatenated_string)

# Part B: Concatenating String 3 to the existing string
string3 = ", Welcome to Python programming."
final_string = concatenated_string + string3
print(final_string)
```

```python
# Task 2: String Slicing and Indexing
# Input string from the previous task
text = "Hello Zara, Welcome to Python programming."

# a. Print the first character of the string
print("a. First character:", text[0])

# b. Print the last character of the string
print("b. Last character:", text[-1])

# c. Print the first 5 characters of the string
print("c. First 5 characters:", text[:5])

# d. Print the last 11 characters of the string
print("d. Last 11 characters:", text[-11:])

# e. Print the string in reverse
print("e. Reversed string:", text[::-1])

# f. Use slicing and print the word "Python" from the existing string
python_index = text.find("Python")
print("f. Extracted word:", text[python_index : python_index + 6])
```

```python
# Task 3: String Methods
strM = "Python beginner tutorial."

# a. Convert the sentence to uppercase
print("a. Uppercase:", strM.upper())

# b. Convert the sentence to lowercase
print("b. Lowercase:", strM.lower())

# c. Use Capitalize and return the sentence to the original input form
print("c. Capitalized:", strM.capitalize())

# d. Count the total number of occurrences of character 't' in the string
print("d. Count of 't':", strM.count("t"))

# e. Replace all occurrences of "Python" with "Data Analytics"
replaced_text = strM.replace("Python", "Data Analytics")
print("e. Replaced string:", replaced_text)
```

```python
# Task 4: Tuples Operations

# Creating the tuples
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate the two tuples and store it in "t_combine"
t_combine = tuple1 + tuple2
print("a. Combined Tuple:", t_combine)

# b. Repeat the elements of "t_combine" 3 times
t_repeated = t_combine * 3
print("b. Repeated 3 times:", t_repeated)

# c. Access the 3rd element from "t_combine"
print("c. 3rd element:", t_combine[2])

# d. Access the first three elements from "t_combine"
print("d. First three elements:", t_combine[:3])

# e. Access the last three elements from "t_combine"
print("e. Last three elements:", t_combine[-3:])
```
